```python
! pip install BeautifulSoup4

from bs4 import BeautifulSoup
# import requests
import urllib.request
import pandas as pd

#[CODE 1]
def lms(result):
    for _ in range(1,9):
        lms_url = 'https://lms.kyonggi.ac.kr/local/ubion/user/'
        print(lms_url)
        html = urllib.request.urlopen(lms_url)
        soup = BeautifulSoup(html, 'html.parser')
        
        # 1. 강의 목록 크롤링
        course_list_divs = soup.select("div.course_lists ul.my-course-lists div.course-title")
        for course_div in course_list_divs:
            course_title = course_div.get_text(strip=True)
            prof = course_div.find_next_sibling("div", class_="prof")
            prof_name = prof.get_text(strip=True) if prof else ""
            result.append([course_title, prof_name])

        # 2. 공지사항 크롤링
        notices = soup.select("div.tab-content ul.notices li")  # li가 있을 수도 있음
        for notice in notices:
            link_tag = notice.find("a")
            link = link_tag['href'] if link_tag else ""
            date_tag = notice.find("div", class_="date")
            date = date_tag.get_text(strip=True) if date_tag else ""
            title_tag = notice.find("h4")
            title = title_tag.get_text(strip=True) if title_tag else ""
            result.append([title, date, link])
    
    return

#[CODE 0]
def main():
    result = []
    print('Crawling >>>>>>>>>>>>>>>>>>>>>>>>>>')
    res = lms(result)   #[CODE 1] 호출 
    lms_table = pd.DataFrame(res, columns=['course_title', 'prof_name', 'title', 'date', 'link'])
    lms_table.to_csv('lmsCrawling.csv', encoding='cp949', mode='w', index=True)
    print('melonCrawl.csv  파일저장 완료 >>>>>>>>>>>>>>>>>')
       
if __name__ == '__main__':
     main()
```
