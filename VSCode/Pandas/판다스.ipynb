```python
data = {
    "국어": [80, 90, 70, 30],
    "영어": [90, 70, 60, 40],
    "수학": [90, 60, 80, 70],
}
columns = ["국어", "영어", "수학"]
index = ["춘향", "몽룡", "향단", "방자"]
df = pd.DataFrame(data, index=index, columns=columns)

df
```

```python
# (1) 모든 학생의 수학 점수를 시리즈로 나타낸다.
math = pd.Series(data["수학"], index = index)
math
# (2) 모든 학생의 국어와 영어 점수를 데이터 프레임으로 나타낸다.
korEng = pd.DataFrame(df.T["국어":"영어"])
korEng
# (3) 모든 학생의 각 과목 평균 점수를 새로운 열로 추가한다.
df["세 과목 평균"] = (df["국어"] + df["수학"] + df["영어"]) / len(columns)
df
# (4) 방자의 영어 점수를 80점으로 수정하고 평균 점수도 다시 계산한다.
df["영어"]["방자"] = 80
df["세 과목 평균"] = round((df["국어"] + df["수학"] + df["영어"]) / len(columns), 2)  #round(식, 2) -> 계산되는 식을 소수점 2번째 자리까지 표사하기
df
# (5) 춘향의 점수를 데이터프레임으로 나타낸다.
choon = pd.DataFrame(df.T["춘향"])
choon
# (6) 향단의 점수를 시리즈로 나타낸다.
hyang = pd.Series(df.T["향단"])
hyang
```

각 승객에 대해 다음 조건을 조합해 risk_group 열을 생성

- 나이age가 15세 이하, 성별 female -> "high-survival"
- 선실 등급class이 First지만 성별이 male -> "mid-survival"
- 그 외 -> "low-survival"

risk_group별 생존율survived을 계산해 어떤 조건이 생존에 가장 유리한 지 분석

### 문제2
다음 두 집단 비교:
1) sibsp == 0
2) sibsp >= 3

- 각 집단의 생존율, 평균 나이, 평균 요금을 비교하여 어떤 차이 있는지 분석해 DataFrame 형태로 출력

```python
titanic["age"] = titanic["age"].fillna("NaN").astype(float)

titanic["risk_group"] = titanic.apply(lambda x: 
                                      "high-survival" if (x.sex == "female" and x.age <= 15) 
                                      else "mid-survival" if (x['class'] == 'First' and x.sex == "male")
                                      else "low-survival", axis=1)

titanic.groupby(titanic["risk_group"])["survived"].mean()
```

```python
group1 = titanic[titanic["sibsp"] == 0]
# titanic.apply(lambda x: x["survived"] if (x[x["sibsp"]] == 0) else "NaN", axis=1)
group2 = titanic[titanic["sibsp"] >= 3]
# titanic.apply(lambda x: x["survived"] if (x[x["sibsp"]] >= 3) else "NaN", axis=1)

avgage = titanic["age"].mean().astype(int)
avgfee = titanic["fare"].mean().astype(int)
gr1 = pd.DataFrame(titanic[group1], avgage, avgfee)
gr2 = pd.DataFrame(titanic[group2], avgage, avgfee)
print(f"{gr1} & {gr2}")
```
