```python
data = {
    "국어": [80, 90, 70, 30],
    "영어": [90, 70, 60, 40],
    "수학": [90, 60, 80, 70],
}
columns = ["국어", "영어", "수학"]
index = ["춘향", "몽룡", "향단", "방자"]
df = pd.DataFrame(data, index=index, columns=columns)

df
```

```python
# (1) 모든 학생의 수학 점수를 시리즈로 나타낸다.
math = pd.Series(data["수학"], index = index)
math
# (2) 모든 학생의 국어와 영어 점수를 데이터 프레임으로 나타낸다.
korEng = pd.DataFrame(df.T["국어":"영어"])
korEng
# (3) 모든 학생의 각 과목 평균 점수를 새로운 열로 추가한다.
df["세 과목 평균"] = (df["국어"] + df["수학"] + df["영어"]) / len(columns)
df
# (4) 방자의 영어 점수를 80점으로 수정하고 평균 점수도 다시 계산한다.
df["영어"]["방자"] = 80
df["세 과목 평균"] = round((df["국어"] + df["수학"] + df["영어"]) / len(columns), 2)  #round(식, 2) -> 계산되는 식을 소수점 2번째 자리까지 표사하기
df
# (5) 춘향의 점수를 데이터프레임으로 나타낸다.
choon = pd.DataFrame(df.T["춘향"])
choon
# (6) 향단의 점수를 시리즈로 나타낸다.
hyang = pd.Series(df.T["향단"])
hyang
```
