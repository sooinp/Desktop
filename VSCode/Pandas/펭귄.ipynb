```python
import pandas as pd
import numpy as np

import seaborn as sns
penguins = sns.load_dataset("penguins")
penguins
```

### 1번 문제 
- 성별 × 종(species)에 따른 평균 체중 비교
- body_mass_g 컬럼을 이용해, 펭귄의 몸무게를 kg 단위로 변환한 body_mass_kg 파생변수를 만들어라. (공식: body_mass_kg = body_mass_g / 1000)
- 성별(sex) × 종(species) 조합별 평균 body_mass_kg를 계산하라.
- 어떤 종 × 성별 조합이 가장 무거운지를 찾고, 그 이유를
- bill_length_mm(부리 길이)
- flipper_length_mm(날개 길이)같은 다른 변수와 연결해서 데이터적으로 해석해보라.

```python
penguins["body_mass_kg"] = penguins["body_mass_g"] / 1000

#성별 * 종 조합별 평균 몸무게(kg)
kgs = penguins.groupby(["sex", "species"])[["body_mass_kg"]].mean().round(3)

#몸무게가 무거운 순으로 Top3 뽑기
kgs.sort_values("body_mass_kg", ascending = False).head(3)
```

```python
#몸무게가 최댓값인 하나만 표출
kgs[kgs["body_mass_kg"] == kgs["body_mass_kg"].max()]
```

```python
#종별 몸무게(kg)에 따른 부리 길이와 날개 길이의 경향성 - Gentoo가 몸무게가 큰 편(그 중에서도  Male > Female), 그만큼 부리도 날개도 큼직큼직해서 상위권을 차지한다.
#그 다음 등장하는 건 아델리
addkgs = penguins.groupby(["sex", "species", "body_mass_kg"])[["bill_length_mm", "flipper_length_mm"]].mean().round(3)
addkgs.sort_values("body_mass_kg", ascending = False).head(45)
```

```python
#각 펭귄 종별 몸무게(kg)와 부리 길이 비교
penguins.groupby(["species", "body_mass_kg"])[["bill_length_mm"]].mean().round(3)
```

```python
#각 펭귄 종별 몸무게(kg)와 날개 길이 비교
penguins.groupby(["species", "body_mass_kg"])[["flipper_length_mm"]].mean().round(3)
```

### 2번 문제 
- flipper 길이 기반 “날개 레벨” 만들기
- flipper_length_mm(날개 길이)를 기준으로 팽귄을 세 구간으로 나누어라.
  - 하위 50% 미만 → "short"
  - 50% ~ 80% 분위수 → "mid"
  - 상위 20% 이상 → "long"
  - 이 기준을 사용해 flipper_level 이라는 파생 변수를 만들어라. (힌트: quantile()과 조건문 또는 pd.qcut() 활용 가능)

- species(종) × island(서식지) × flipper_level 조합별로 평균 body_mass_g를 계산하라.

- 평균 body_mass_g가 가장 높은 조합 TOP 3를 내림차순으로 출력하라.

TOP 1 조합이 왜 무거운지,
해당 종의 특성
해당 island(서식지) 환경
flipper_level(날개 길이)를 바탕으로 데이터적 관점에서 해석해보라.

```python
penguins["flipper_level"] = pd.qcut(penguins["flipper_length_mm"], q = [0, 0.5, 0.8, 1.0], labels = ["short", "mid", "long"])

#종 * 서식지 * 날개 레벨 조합별로 평균 몸무게(g) 계산
penguins.groupby(["species", "island", "flipper_level"])[["body_mass_g"]].mean().round(3)
##아델리는 날개가 짧은 편이다./신스트랩은 드림에서만 산다./젠투는 비스코에서만 산다. + 날개도 긴 편
```

```python
penguins[["body_mass_g"]].fillna(0.0)
penguins.groupby(["species", "island", "flipper_level"])["body_mass_g"].mean().sort_values(ascending = False).head(3)
```

### 3번 문제 
- “일반적으로 무거운 펭귄 그룹” 찾기
- 다음 네 가지 조건을 모두 사용하는 그룹을 만들라:
  - 종(species)
  - 섬(island)
  - 성별(sex)
- 날개 레벨(flipper_level, 2번 문제에서 만든 것) 즉, species × island × sex × flipper_level 조합으로 그룹을 만든다.

- 각 그룹에 대해 body_mass_g의 평균(mean), 분산(var)을 동시에 구하라.

- 평균 몸무게가 크면서, 분산은 낮은 그룹을 “일반적으로 무거운 그룹”이라고 정의
- 이 기준에 따라 TOP 3 그룹을 선정하라. (예: score = mean / (var + 작은 값) 처럼 점수를 만들어도 됨)
- 선택된 TOP 3 그룹의 공통점이 무엇인지 종(species), 성별(sex), flipper_level(날개 길이) 관점에서 해석해보라.

```python
heavy_penguins = penguins.groupby(["species", "island", "sex", "flipper_level"])["body_mass_g"].agg(["mean", "var"])

heavy_penguins.sort_values(by=["mean", "var"], ascending=[False, True]).round(3)
```
