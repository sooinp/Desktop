```python
# 정답 :))
import numpy as np
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [5, 3, 6, 4]
y2 = [4, 4, 5, 5]

# 리스트를 numpy 배열로 변환 (조건 비교를 위해)
x = np.array(x)
y1 = np.array(y1)
y2 = np.array(y2)

plt.plot(x, y1, label='y1')
plt.plot(x, y2, label='y2')

# y1 > y2 인 구간 → 초록색
plt.fill_between(
    x, y1, y2,
    where=(y1 > y2),
    color='green',
    alpha=0.5,
    interpolate=True
)

# y1 < y2 인 구간 → 빨간색
plt.fill_between(
    x, y1, y2,
    where=(y1 < y2),
    color='red',
    alpha=0.5,
    interpolate=True
)

plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.legend()
plt.show()
```

```python
# 연습 문제 
import numpy as np
import matplotlib.pyplot as plt

# 데이터 생성
x = np.arange(0, 10, 1)
y = x ** 2

# 기본 그래프
plt.plot(x, y, marker='o', label='y = x²')

# 그리드 (수직 + 수평)
plt.grid(True, axis='both', color='gray', alpha=0.3, linestyle='--')

# 수평선 (기준값 표시)
plt.axhline(y=25, color='red', linestyle='--', linewidth=2, label='y = 25 기준선')

# 수직선 (특정 x값 강조)
plt.axvline(x=5, color='blue', linestyle=':', linewidth=2, label='x = 5 기준선')

# 옵션
plt.xlabel('x')
plt.ylabel('y')
plt.legend()

plt.show()
```

```python
# 예제 : 도넛 차트를 그리

import numpy as np
import matplotlib.pyplot as plt

# 1. 원본 판매 데이터
products = [
    'Apple', 'Banana', 'Apple', 'Melon', 'Grapes',
    'Banana', 'Apple', 'Melon', 'Apple', 'Grapes',
    'Banana', 'Banana', 'Apple', 'Melon', 'Grapes',
    'Apple', 'Apple', 'Banana', 'Melon', 'Apple'
]

# 2. 상품별 판매 개수 집계
items, counts = np.unique(products, return_counts=True)

# 3. 판매량 3개 이하 → Other로 묶기
labels = []
final_counts = []

other_count = 0
for item, count in zip(items, counts):
    if count <= 3:
        other_count += count
    else:
        labels.append(item)
        final_counts.append(count)

if other_count > 0:
    labels.append('Other')
    final_counts.append(other_count)

labels = np.array(labels)
final_counts = np.array(final_counts)

# 4. 비율 계산
ratios = final_counts / final_counts.sum() * 100

# 5. 가장 많이 팔린 상품 explode 처리
explode = [0.1 if count == final_counts.max() else 0 for count in final_counts]

# 6. 도넛 차트 스타일 설정
colors = ['#ff9999', '#ffc000', '#8fd9b6', '#d395d0', '#c2c2f0']

wedgeprops = {
    'width': 0.4,        # 도넛 두께
    'edgecolor': 'w',
    'linewidth': 2
}

# 7. 도넛 파이 차트 그리기
plt.figure(figsize=(6, 6))
plt.pie(
    ratios,
    labels=labels,
    autopct='%.1f%%',
    startangle=90,
    counterclock=False,
    colors=colors[:len(labels)],
    explode=explode,
    wedgeprops=wedgeprops
)

plt.title('Product Sales Ratio (Donut Chart)')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


# 1) 계층 구조
# 바깥: 3그룹(과자/과일/음료)
# 안쪽: 각 그룹 2개씩(세부항목)
outer_labels = ["과자", "과일", "음료"]
inner_labels = [
    ["쿠키", "칩"],
    ["사과", "바나나"],
    ["주스", "우유"]
]
vals = np.array([
    [30, 20],  # 과자: 쿠키 30, 칩 20  -> 합 50
    [25, 15],  # 과일: 사과 25, 바나나 15 -> 합 40
    [20, 10]   # 음료: 주스 20, 우유 10  -> 합 30
], dtype=float)


# 2) 파이(각도)로 바꾸기: 전체 합 기준으로 2π
total = vals.sum()
valsnorm = vals / total * 2 * np.pi  # 각 조각의 각도(라디안)

# 각 조각이 "어디서 시작하는지" 누적합으로 계산
flat = valsnorm.flatten()
starts_flat = np.concatenate([[0], np.cumsum(flat[:-1])])  # 첫 조각 0부터
starts = starts_flat.reshape(vals.shape)


# 3) 극좌표(Polar) 세팅
fig, ax = plt.subplots(figsize=(8, 6), subplot_kw=dict(projection="polar"))
ax.set_title("도넛 파이(극좌표 + bar)", pad=20)

ring = 0.25  # 고리 두께

# 색상(단순하게 3그룹/6항목)
outer_colors = ["#4C72B0", "#55A868", "#C44E52"]  # 파랑/초록/빨강 느낌
inner_colors = ["#8FBCE6", "#2E5EAA", "#8AD19D", "#2E8B57", "#E79A9A", "#B23B3B"]

# 4) 바깥 고리(그룹) 그리기
outer_widths = valsnorm.sum(axis=1)     # 그룹별 총 각도
outer_starts = starts[:, 0]             # 각 그룹의 시작 각도(첫 하위항목 시작점)

ax.bar(
    x=outer_starts,
    width=outer_widths,
    bottom=1 - ring,
    height=ring,
    color=outer_colors,
    edgecolor="white",
    linewidth=1,
    align="edge"
)

# 바깥 고리 글자(그룹명 + 합)
for i in range(len(outer_labels)):
    theta_center = outer_starts[i] + outer_widths[i] / 2
    r_center = 1 - ring/2
    group_sum = int(vals[i].sum())
    ax.text(theta_center, r_center, f"{outer_labels[i]}\n{group_sum}",
            ha="center", va="center", fontsize=10, fontweight="bold")

# 5) 안쪽 고리(세부항목) 그리기
ax.bar(
    x=starts.flatten(),
    width=valsnorm.flatten(),
    bottom=1 - 2*ring,
    height=ring,
    color=inner_colors,
    edgecolor="white",
    linewidth=1,
    align="edge"
)

# 안쪽 고리 글자(세부항목명 + 값)
k = 0
for i in range(vals.shape[0]):        # 그룹
    for j in range(vals.shape[1]):    # 세부항목
        theta_center = starts[i, j] + valsnorm[i, j] / 2
        r_center = 1 - 1.5*ring
        ax.text(theta_center, r_center, f"{inner_labels[i][j]}\n{int(vals[i,j])}",
                ha="center", va="center", fontsize=9)
        k += 1

# 보기 좋게 축 숨김(파이 느낌)
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

#### 5-3) Custom hillshading in a 3D surface plot

```python
import matplotlib.pyplot as plt
import numpy as np

from matplotlib import cbook, cm
from matplotlib.colors import LightSource

# Load and format data
dem = cbook.get_sample_data('jacksboro_fault_dem.npz')
z = dem['elevation']
nrows, ncols = z.shape
x = np.linspace(dem['xmin'], dem['xmax'], ncols)
y = np.linspace(dem['ymin'], dem['ymax'], nrows)
x, y = np.meshgrid(x, y)

region = np.s_[5:50, 5:50]
x, y, z = x[region], y[region], z[region]

# Set up plot
fig, ax = plt.subplots(subplot_kw=dict(projection='3d'))

ls = LightSource(270, 45)
# To use a custom hillshading mode, override the built-in shading and pass
# in the rgb colors of the shaded surface calculated from "shade".
rgb = ls.shade(z, cmap=cm.gist_earth, vert_exag=0.1, blend_mode='soft')
surf = ax.plot_surface(x, y, z, rstride=1, cstride=1, facecolors=rgb,
                       linewidth=0, antialiased=False, shade=False)

plt.show()
```

#### 5-4) Fill under 3D line graphs

```python
import math

import matplotlib.pyplot as plt
import numpy as np

gamma = np.vectorize(math.gamma)
N = 31
x = np.linspace(0., 10., N)
lambdas = range(1, 9)

ax = plt.figure().add_subplot(projection='3d')

facecolors = plt.colormaps['viridis_r'](np.linspace(0, 1, len(lambdas)))

for i, l in enumerate(lambdas):
    # Note fill_between can take coordinates as length N vectors, or scalars
    ax.fill_between(x, l, l**x * np.exp(-l) / gamma(x + 1),
                    x, l, 0,
                    facecolors=facecolors[i], alpha=.7)

ax.set(xlim=(0, 10), ylim=(1, 9), zlim=(0, 0.35),
       xlabel='x', ylabel=r'$\lambda$', zlabel='probability')

plt.show()
```
