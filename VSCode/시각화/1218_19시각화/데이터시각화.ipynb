```python
# 문제 : 차이에 따라 색 다르게 채우기
# 아래 데이터에서
# y1 > y2 인 구간은 초록색
# y1 < y2 인 구간은 빨간색으로 채워주세요 >.<

x = [1, 2, 3, 4]
y1 = [5, 3, 6, 4]
y2 = [4, 4, 5, 5]

plt.plot(x, y1, label = 'y1' ) #파란색
plt.plot(x, y2, label = 'y2') #빨간색
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.legend(loc='upper right', ncol=1, fontsize=8)

new_x, new_y1, new_y2 = [], [], []
for i in range(len(x) - 1):
    # 현재 점 추가
    new_x.append(x[i])
    new_y1.append(y1[i])
    new_y2.append(y2[i])
    
    # 교차 여부 확인 (y1-y2의 부호가 바뀌면 교차)
    if (y1[i] - y2[i]) * (y1[i+1] - y2[i+1]) < 0:
        # 직선 방정식 연립: x = x1 + (x2-x1) * |d1| / (|d1| + |d2|)
        d1 = y1[i] - y2[i]
        d2 = y1[i+1] - y2[i+1]
        inter_x = x[i] + (x[i+1] - x[i]) * abs(d1) / (abs(d1) + abs(d2))
        inter_y = y1[i] + (y1[i+1] - y1[i]) * abs(d1) / (abs(d1) + abs(d2))
        
        new_x.append(inter_x)
        new_y1.append(inter_y)
        new_y2.append(inter_y) # 교차점이므로 y1 = y2
        
# 마지막 점 추가
new_x.append(x[-1])
new_y1.append(y1[-1])
new_y2.append(y2[-1])

for i in range(len(new_x) - 1):
    section_x = [new_x[i], new_x[i+1]]
    section_y1 = [new_y1[i], new_y1[i+1]]
    section_y2 = [new_y2[i], new_y2[i+1]]
    
    # 구간의 중앙값으로 색상 판별
    mid_y1 = (new_y1[i] + new_y1[i+1]) / 2
    mid_y2 = (new_y2[i] + new_y2[i+1]) / 2
    
    colors = 'green' if mid_y1 > mid_y2 else 'red'
    
    plt.fill_between(section_x, section_y1, section_y2, color=colors, alpha=0.5)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt


# 1) 계층 구조
# 바깥: 3그룹(과자/과일/음료)
# 안쪽: 각 그룹 2개씩(세부항목)
outer_labels = ["과자", "과일", "음료"]
inner_labels = [
    ["쿠키", "칩"],
    ["사과", "바나나"],
    ["주스", "우유"]
]
vals = np.array([
    [30, 20],  # 과자: 쿠키 30, 칩 20  -> 합 50
    [25, 15],  # 과일: 사과 25, 바나나 15 -> 합 40
    [20, 10]   # 음료: 주스 20, 우유 10  -> 합 30
], dtype=float)


# 2) 파이(각도)로 바꾸기: 전체 합 기준으로 2π
total = vals.sum()
valsnorm = vals / total * 2 * np.pi  # 각 조각의 각도(라디안)

# 각 조각이 "어디서 시작하는지" 누적합으로 계산
flat = valsnorm.flatten()
starts_flat = np.concatenate([[0], np.cumsum(flat[:-1])])  # 첫 조각 0부터
starts = starts_flat.reshape(vals.shape)


# 3) 극좌표(Polar) 세팅
fig, ax = plt.subplots(figsize=(8, 6), subplot_kw=dict(projection="polar"))
ax.set_title("도넛 파이(극좌표 + bar)", pad=20)

ring = 0.25  # 고리 두께

# 색상(단순하게 3그룹/6항목)
outer_colors = ["#4C72B0", "#55A868", "#C44E52"]  # 파랑/초록/빨강 느낌
inner_colors = ["#8FBCE6", "#2E5EAA", "#8AD19D", "#2E8B57", "#E79A9A", "#B23B3B"]

# 4) 바깥 고리(그룹) 그리기
outer_widths = valsnorm.sum(axis=1)     # 그룹별 총 각도
outer_starts = starts[:, 0]             # 각 그룹의 시작 각도(첫 하위항목 시작점)

ax.bar(
    x=outer_starts,
    width=outer_widths,
    bottom=1 - ring,
    height=ring,
    color=outer_colors,
    edgecolor="white",
    linewidth=1,
    align="edge"
)

# 바깥 고리 글자(그룹명 + 합)
for i in range(len(outer_labels)):
    theta_center = outer_starts[i] + outer_widths[i] / 2
    r_center = 1 - ring/2
    group_sum = int(vals[i].sum())
    ax.text(theta_center, r_center, f"{outer_labels[i]}\n{group_sum}",
            ha="center", va="center", fontsize=10, fontweight="bold")

# 5) 안쪽 고리(세부항목) 그리기
ax.bar(
    x=starts.flatten(),
    width=valsnorm.flatten(),
    bottom=1 - 2*ring,
    height=ring,
    color=inner_colors,
    edgecolor="white",
    linewidth=1,
    align="edge"
)

# 안쪽 고리 글자(세부항목명 + 값)
k = 0
for i in range(vals.shape[0]):        # 그룹
    for j in range(vals.shape[1]):    # 세부항목
        theta_center = starts[i, j] + valsnorm[i, j] / 2
        r_center = 1 - 1.5*ring
        ax.text(theta_center, r_center, f"{inner_labels[i][j]}\n{int(vals[i,j])}",
                ha="center", va="center", fontsize=9)
        k += 1

#한글 깨지면 이거 무조건 넣어야 됨
plt.rc('font',family='Malgun Gothic')

# 보기 좋게 축 숨김(파이 느낌)
ax.set_axis_off()
plt.tight_layout()
plt.show()
```

```python
#이미지 사이즈 디폴트 값으로 복구
plt.rcdefaults()

#  make data:
np.random.seed(1)
x = [2,4,6]
#0 부터 10 사이의 숫자를 그룹마다 50개씩 생성
D = np.random.uniform(0, 10, size=(3, 50))

#plot:
fig, ax = plt.subplots()

#산점도와 차이나는 부분: 산점도는 점으로 데이터 표시한다면 이벤트 플롯은 짧은 선으로 표시
ax.eventplot(D, orientation='vertical', lineoffsets=x, linewidths=0.75)

ax.set_xlim(0, 8)
ax.set_xticks(x)
ax.set_xticklabels(['Group A', 'Group B', 'Group C'])

ax.set_ylim(0, 11)
ax.set_yticks(np.arange(0, 11, 1))

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

#난수 고정(실습용)
np.random.seed(42)

#그룹 개수
num_groups  = 5

#각 그룹당 이벤트 개수
events_per_group = 30

#랜덤 데이터 생성(0 ~ 10)
data = [np.random.uniform(0, 10, events_per_group) for _ in range(num_groups)]

#각 그룹의 y 위치
y_positions = np.arange(1, num_groups + 1)

#색상 리스트(그룹별 색)
colors = ['red', 'blue', 'green', 'orange', 'purple']

#Figure & Axes
fig, ax = plt.subplots(figsize = (10, 5))

# 이벤트 플롯 (가로 방향)
ax.eventplot(
    data,
    orientation='horizontal', #vertical 로 바꾸면 방향 바뀜
    lineoffsets=y_positions,
    colors=colors,
    linewidths=1
)

# 축 설정
ax.set(
    xlabel='Event Value',
    ylabel='Group',
    yticks=y_positions,
    yticklabels=[f'Group {i}' for i in y_positions],
    xlim=(0, 10)
)

plt.title('Event Plot (5 Groups, Random Data)')
plt.tight_layout()
plt.show()
```

## 3) 여러 개의 그래프

```python
x1 = np.linspace(0.0, 5.0)
x2 = np.linspace(0.0, 2.0)

y1 = np.cos(2 * np.pi * x1) * np.exp(-x1)   #코사인감쇠진동함수
y2 = np.cos(2 * np.pi * x2)                 #코사인함수

#subplot() 위치들만 확인
plt.subplot(2, 1, 1) #nrows = 2, ncols = 1, index = 1
plt.plot(x1, y1, 'o-')
plt.title('1st Graph')
plt.ylabel("Damped oscillation")

plt.subplot(2, 1, 2) #nrows = 2, ncols = 1, index = 2
plt.plot(x2, y2, '.-')
plt.title('2nd Graph')
plt.xlabel("time (s)")
plt.ylabel("Undamped")

plt.tight_layout()
plt.show()
```

```python
a = 2.0 * np.random.randn(10000) + 1.0
b = np.random.standard_normal(10000)
c = 20.0 * np.random.rand(5000) - 10.0

font1 = {'family': 'serif',
        'color':  'darkred',
        'weight': 'normal',
        'size': 16}

font2 = {'family': 'Times New Roman',
        'color':  'blue',
        'weight': 'bold',
        'size': 12,
        'alpha': 0.7}

font3 = {'family': 'Arial',
        'color':  'forestgreen',
        'style': 'italic',
        'size': 14}

box1 = {'boxstyle': 'round',
        'ec': (1.0, 0.5, 0.5),
        'fc': (1.0, 0.8, 0.8)}

box2 = {'boxstyle': 'square',
        'ec': (0.5, 0.5, 1.0),
        'fc': (0.8, 0.8, 1.0),
        'linestyle': '--'}

box3 = {'boxstyle': 'square',
        'ec': (0.3, 1.0, 0.5), #‘ec’는 ‘edgecolor’
        'fc': (0.8, 1.0, 0.5), #‘fc’는 ‘facecolor’
        'linestyle': '-.',
        'linewidth': 2}

plt.hist(a, bins=100, density=True, alpha=0.7, histtype='step')
plt.text(-3.0, 0.15, 'np.random.randn()', fontdict=font1, rotation=85, bbox=box1)
plt.hist(b, bins=50, density=True, alpha=0.5, histtype='stepfilled')
plt.text(2.0, 0.0, 'np.random.standard_normal()', fontdict=font2, rotation=-60, bbox=box2)
plt.hist(c, bins=100, density=True, alpha=0.9, histtype='step')
plt.text(-10.0, 0.08, 'np.random.rand()', fontdict=font3, bbox=box3)
plt.show()
```
