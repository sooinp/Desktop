```python
# 문제 : 차이에 따라 색 다르게 채우기
# 아래 데이터에서
# y1 > y2 인 구간은 초록색
# y1 < y2 인 구간은 빨간색으로 채워주세요 >.<

x = [1, 2, 3, 4]
y1 = [5, 3, 6, 4]
y2 = [4, 4, 5, 5]

plt.plot(x, y1, label = 'y1' ) #파란색
plt.plot(x, y2, label = 'y2') #빨간색
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.legend(loc='upper right', ncol=1, fontsize=8)

new_x, new_y1, new_y2 = [], [], []
for i in range(len(x) - 1):
    # 현재 점 추가
    new_x.append(x[i])
    new_y1.append(y1[i])
    new_y2.append(y2[i])
    
    # 교차 여부 확인 (y1-y2의 부호가 바뀌면 교차)
    if (y1[i] - y2[i]) * (y1[i+1] - y2[i+1]) < 0:
        # 직선 방정식 연립: x = x1 + (x2-x1) * |d1| / (|d1| + |d2|)
        d1 = y1[i] - y2[i]
        d2 = y1[i+1] - y2[i+1]
        inter_x = x[i] + (x[i+1] - x[i]) * abs(d1) / (abs(d1) + abs(d2))
        inter_y = y1[i] + (y1[i+1] - y1[i]) * abs(d1) / (abs(d1) + abs(d2))
        
        new_x.append(inter_x)
        new_y1.append(inter_y)
        new_y2.append(inter_y) # 교차점이므로 y1 = y2
        
# 마지막 점 추가
new_x.append(x[-1])
new_y1.append(y1[-1])
new_y2.append(y2[-1])

for i in range(len(new_x) - 1):
    section_x = [new_x[i], new_x[i+1]]
    section_y1 = [new_y1[i], new_y1[i+1]]
    section_y2 = [new_y2[i], new_y2[i+1]]
    
    # 구간의 중앙값으로 색상 판별
    mid_y1 = (new_y1[i] + new_y1[i+1]) / 2
    mid_y2 = (new_y2[i] + new_y2[i+1]) / 2
    
    colors = 'green' if mid_y1 > mid_y2 else 'red'
    
    plt.fill_between(section_x, section_y1, section_y2, color=colors, alpha=0.5)

plt.show()
```
