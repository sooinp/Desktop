```python
from bs4 import BeautifulSoup
import urllib.request
import pandas as pd
import datetime
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
import time
from selenium.common.exceptions import UnexpectedAlertPresentException, NoAlertPresentException
```

```python
def setup_driver():
    """로그인 회피 및 봇 탐지 방지 설정"""
    chrome_options = Options()
    # 자동화 메시지 제거
    chrome_options.add_experimental_option("excludeSwitches", ["enable-automation"])
    chrome_options.add_experimental_option('useAutomationExtension', False)
    # 일반 사용자처럼 보이게 하는 User-Agent 설정
    chrome_options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36")
    
    driver = webdriver.Chrome(options=chrome_options)
    # webdriver-manager 속성을 제거하여 봇 감지 회피
    driver.execute_cdp_cmd("Page.addScriptToEvaluateOnNewDocument", {
        "source": """
            Object.defineProperty(navigator, 'webdriver', {
                get: () => undefined
            })
        """
    })
    return driver
```

```python
#[CODE 1]
def Yes24(bookName):
    # 크롬 WebDriver 객체 생성
    wd = webdriver.Chrome()
    # 특정 페이지 이동
    wd.get("https://www.yes24.com/main/")
    all_book = []

    for book in bookName:
        try:
            # 정보 분리 (도서명/저자/출판사)
            parts = book.split("/")
            titles = parts[0]
            author = parts[1] if len(parts) > 1 else ""
            publisher = parts[2] if len(parts) > 2 else ""            
            
            print(f"'{titles}' 검색 중...")
            search_box = wd.find_element(By.ID, "query")
            search_box.clear() # 이전 검색어 삭제
            search_box.send_keys(titles) # 책 제목 입력
            search_box.send_keys(Keys.RETURN)
            time.sleep(1) #스크립트 실행 할 동안 1초 대기
            
            # 검색 결과 리스트에서 일치하는 책 찾기
            items = wd.find_elements(By.CSS_SELECTOR, "#yesSchList > li")
            target_link = None
            
            for item in items:
                try:
                    # 검색 결과 내 저자와 출판사 텍스트 확인
                    info_text = item.find_element(By.CLASS_NAME, "info_pub").text
                    
                    # 저자와 출판사가 포함되어 있는지 확인
                    if author in info_text and publisher in info_text:
                        target_link = item.find_element(By.CLASS_NAME, "gd_name")
                        break
                except:
                    continue
            
            # 일치하는 항목이 있으면 클릭, 없으면 첫 번째 항목 클릭
            if target_link:
                target_link.click()
            else:
                print("일치 항목 없음: 첫 번째 결과 선택")
                wd.find_element(By.CSS_SELECTOR, "#yesSchList > li:nth-child(1) a.gd_name").click()
            
            # --- [핵심 수정] 성인 인증 경고창(Alert) 처리 ---
            try:
                alert = wd.switch_to.alert
                alert_text = alert.text
                print(f"!!! 성인 인증 필요 도서 건너뜀: {alert_text}")
                alert.accept() # 확인 버튼 누름
                wd.get("https://www.yes24.com/main/") # 메인으로 이동
                continue # 다음 책으로 이동
            except NoAlertPresentException:
                # 경고창이 없으면 정상 진행
                pass
            
            # 책 상세 페이지에서 정보 추출
            html = wd.page_source
            soup = BeautifulSoup(html, 'html.parser')
            
            book_data = {
                '도서명': soup.select_one(".gd_name").text.strip() if soup.select_one(".gd_name") else titles,
                '저자': soup.select_one(".gd_auth").text.strip() if soup.select_one(".gd_auth") else None,
                '출판사': soup.select_one(".gd_pub").text.strip() if soup.select_one(".gd_pub") else None,
                '발행년도': soup.select_one(".gd_date").text.strip()[:4] if soup.select_one(".gd_date") else None,
                'ISBN': None, '부가기호': None, '주제분류번호': None
            }
            
            # 상세 정보 섹션 파싱
            info_wrap = soup.select_one("#infoset_specific")
            if info_wrap:
                dls = info_wrap.select("dl.gd_infoSet")
                for dl in dls:
                    dt_text = dl.select_one("dt").text.strip()
                    dd_text = dl.select_one("dd").text.strip()
                    
                    if "ISBN" in dt_text:
                        book_data['ISBN'] = dd_text.split("(")[0].strip()
                    elif "부가기호" in dt_text:
                        book_data['부가기호'] = dd_text
                    elif "주제분류" in dt_text:
                        book_data['주제분류번호'] = dd_text

            all_book.append(book_data)
            print(f"추출 완료: {book_data['도서명']}")
            
            # # 저자
            # author_tag = soup.select_one("span.gd_pubArea > span.gd_auth")
            # if author_tag:
            #     book_data['저자'] = author_tag.text.strip()
            # else:
            #     book_data['저자'] = None
            
            # # 출판사
            # publisher_tag = soup.select_one("span.gd_pubArea > a.gd_pub")
            # if publisher_tag:
            #     book_data['출판사'] = publisher_tag.text.strip()
            # else:
            #     book_data['출판사'] = None
                
            # # 발행년도 (출판사 정보 옆에 같이 있는 경우가 많음)
            # pub_date_tag = soup.select_one("span.gd_pubArea > span.gd_date")
            # if pub_date_tag:
            #     book_data['발행년도'] = pub_date_tag.text.strip().split('.')[0] # 'YYYY년 MM월 DD일' -> YYYY
            # else:
            #     book_data['발행년도'] = None
                
            # # '상품정보' 섹션에서 상세 정보 추출 (예시)
            # info_table = soup.select_one("#infoset_specific > div.infoSetCont_wrap > div.gd_infoSet_maker")
            # if info_table:
            #     rows = info_table.select("dl.gd_infoSet > dt, dl.gd_infoSet > dd")
            #     for i in range(0, len(rows), 2):
            #         if i + 1 < len(rows):
            #             key = rows[i].text.strip()
            #             value = rows[i+1].text.strip()

            #             if "ISBN" in key:
            #                 book_data['ISBN'] = value.split()[0] # ISBN13만 추출
            #             elif "부가기호" in key:
            #                 book_data['부가기호'] = value
            #             elif "주제분류" in key:
            #                 book_data['주제분류번호'] = value
                            
            # # 정보가 없는 경우 None으로 처리
            # if 'ISBN' not in book_data: book_data['ISBN'] = None
            # if '부가기호' not in book_data: book_data['부가기호'] = None
            # if '주제분류번호' not in book_data: book_data['주제분류번호'] = None

            # all_book.append(book_data)
            # print(f"'{title}' 정보 추출 완료.")
            
            wd.get("https://www.yes24.com/main/")
            time.sleep(1)
            
        except UnexpectedAlertPresentException as e:
            # 예상치 못한 시점에 경고창이 떴을 때 한 번 더 방어
            print(f"!!! 경고창 감지로 인한 패스: {e.alert_text}")
            try:
                wd.switch_to.alert.accept()
            except: pass
            wd.get("https://www.yes24.com/main/")
            continue    
        
        except Exception as e:
            print(f"'{titles}' 처리 중 오류 발생: {e}")
            wd.get("https://www.yes24.com/main/")
            continue 
    wd.quit() # 모든 작업 완료 후 WebDriver 종료
    return all_book

#[CODE 0]
def main():
    result = []
    print('yes24 crawling >>>>>>>>>>>>>>>>>>>>>>>>>>')
    Yes24(bookName)  #[CODE 1]
    
    BookNames = pd.DataFrame(result, columns=('도서명', '저자', '출판사', '발행년도', 'ISBN', '부가기호', '주제분류번호'))
    BookNames.to_csv('yes24Crawling.csv', encoding='cp949', mode='w', index=True)
    print('파일저장 완료 >>>>>>>>>>>>>>>>>')

if __name__ == '__main__':
     main()
```
