```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([2, 4, 6, 8])
y = np.array([81, 93, 91, 97])

mx = np.mean(x)
my = np.mean(y)

dis = sum([(i - mx)**2 for i in x])

def top(x, mx, y, my):
    d = 0
    for i in range(len(x)):
        d += (x[i] - mx) * (y[i] - my)
        return d

div = top(x, mx, y, my)

a = div / dis
b = my - (mx * a)

print("x의 평균값: ", mx)
print("y의 평균값: ", my)
print("분모: ", dis)
print("분자: ", div)
print("기울기 a= ", a)
print("y절편 b= ", b)
```

```python
fake_a = 3
fake_b = 76

def pred(x):
    return fake_a * x + fake_b

res = []

for i in range(len(x)):
    res.append(pred(x[i]))
    print("공부시간 = %.f, 실제 점수 = %.f, 예측점수 = %.f" % (x[i], y[i], pred(x[i])))

n = len(x)
def mse(y, y_pred):
    return (1/n) * sum((y - y_pred)**2)

print("평균 제곱 오차: ", str(mse(y, res)))
```

```python
plt.scatter(x, y)
plt.show()
```

```python
y_pred = a * x + b
err = y - y_pred

a_dif = (2/n) * sum(-x * (err))
b_dif = (2/n) * sum(-(err))

lr = 0.03 #learning rate
a = a - lr * a_dif
b = b - lr * b_dif
```

```python
a = 0
b = 0
lr = 0.03
epoch = 2001 #에포크, 반복 횟수
n = len(x)

for i in range(epoch):
    y_pred = a * x + b
    err = y - y_pred

    a_dif = (2/n) * sum(-x * (err))
    b_dif = (2/n) * sum(-(err))

    a = a - lr * a_dif
    b = b - lr * b_dif

    if i % 100 == 0:
        print("epoch = %.f, 기울기 = %.04f, 절편 = %.04f" % (i, a, b))

y_pred = a * x + b

plt.scatter(x, y)
plt.plot(x, y_pred, 'r')
plt.show()  
```

다중 선형 회귀

```python
x1 = np.array([2,4,6,8])
x2 = np.array([0,4,2,3])
y = np.array([81,93,91,97])

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.scatter3D(x1, x2, y)
plt.show()
```

```python
a1 = 0
a2 = 0
b = 0
lr = 0.01
epoch = 2001
n = len(x1)

for i in range(epoch):
    y_pred = a1 * x1 + a2 * x2 + b
    err = y - y_pred


    a1_dif = (2/n) * sum(-x1 * (err))
    a2_dif = (2/n) * sum(-x2 * (err))
    b_dif = (2/n) * sum(-(err))

    a1 = a1 - lr * a1_dif
    a2 = a2 - lr * a2_dif
    b = b - lr * b_dif

    if i % 100 == 0:
        print("epoch = %.f, 기울기1 = %.04f, 기울기2 = %.04f, 절편 = %.04f" % (i, a1, a2, b))

print("실제 점수: ", y)
print("예측 점수: ", y_pred)
```
