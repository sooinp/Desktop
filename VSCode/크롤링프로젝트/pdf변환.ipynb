step 1

```python
import shutil

import pdfplumber
import os
```

```python
SOURCE_DIR = "C:/Users/user/OneDrive/Desktop/Desktop/VSCode/크롤링프로젝트"   # PDF들이 흩어져 있는 최상위 폴더: 크롤링프로젝트
PDF_DIR = "pdf모음"           # 모을 폴더

os.makedirs(PDF_DIR, exist_ok=True)

for root, dirs, files in os.walk(SOURCE_DIR):
    for file in files:
        if file.lower().endswith(".pdf"):
            src = os.path.join(root, file)
            dst = os.path.join(PDF_DIR, file)

            # 같은 이름 파일 충돌 방지
            if os.path.exists(dst):
                name, ext = os.path.splitext(file)
                i = 1
                while os.path.exists(dst):
                    dst = os.path.join(PDF_DIR, f"{name}_{i}{ext}")
                    i += 1

            shutil.copy2(src, dst)

print("PDF 수집 완료")
```

```python
PDF_DIR = "pdf모음"
OUTPUT = "merged.txt"

with open(OUTPUT, "w", encoding="utf-8") as out:
    for file in os.listdir(PDF_DIR):
        if file.endswith(".pdf"):
            path = os.path.join(PDF_DIR, file)
            out.write(f"\n\n===== < {file} > =====\n")
            
            with pdfplumber.open(path) as pdf:
                for page in pdf.pages:
                    text = page.extract_text()
                    if text:
                        out.write(text + "\n")
```

```python
from pdf2image import convert_from_path
import pytesseract
```

```python
# pytesseract 사용을 위한 코드 추가
pytesseract.pytesseract.tesseract_cmd = r'C:/Program Files/Tesseract-OCR/tesseract.exe'
```

```python
images = convert_from_path("sample.pdf")

text_all = ""
for img in images:
    text = pytesseract.image_to_string(img, lang="kor+eng")
    text_all += text + "\n"

with open("ocr_result.txt", "w", encoding="utf-8") as f:
    f.write(text_all)
```
