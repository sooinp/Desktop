```python
df['kr_date'] = df['kr_date'].astype(str)
df['us_date'] = df['us_date'].astype(str)
df['kr_year'] = df['kr_date'].str.split('-').str[0]
df['kr_month'] = df['kr_date'].str.split('-').str[1]
df['us_year'] = df['us_date'].str.split('-').str[0]
df['us_month'] = df['us_date'].str.split('-').str[1]
```

```python
# Fixing random state for reproducibility
# np.random.seed(19680802)

domain_colors = {
    '육상/지상': '#1f77b4',   # 파랑
    '항공/공군':     '#ff7f0e',   # 주황
    '해상/해군':       '#2ca02c',   # 초록
    '신기술':  '#d62728',   # 빨강
    '통신/전자':  '#9467bd',   # 보라
    '미사일/방공':  '#8c564b',   # 갈색
    '우주':  '#e377c2',   # 분홍
}
 
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

years = df['kr_year'].astype(int)
df['kr_year_5'] = (years // 5) * 5

df['domain_copy'] = df['domain']
df['domain_copy'] = df['domain_copy'].astype('category')
df['domain_copy'] = df['domain_copy'].cat.reorder_categories(
    df['domain_copy'].cat.categories[::-1],
    ordered=True
)

domain_cats = df['domain_copy'].cat.categories.tolist()
domain_color_list = [domain_colors[d] for d in domain_cats]

x = df['domain_copy'].cat.codes
y = df['kr_year_5'].astype('category').cat.codes
hist, xedges, yedges = np.histogram2d(x, y, bins=[df['domain'].nunique(), df['kr_year_5'].nunique()])

# Construct arrays for the anchor positions of the 16 bars.
xpos, ypos = np.meshgrid(xedges[:-1], yedges[:-1], indexing="ij")
xpos = xpos.ravel()
ypos = ypos.ravel()
zpos = np.zeros_like(xpos)

# Construct arrays with the dimensions for the 16 bars.
dx = dy = 0.8 * np.ones_like(zpos)
dz = hist.ravel()

colors = []
for x_val in xpos:
    idx = int(round(x_val)) # x좌표를 인덱스로 변환
    colors.append(domain_color_list[idx])

ax.set_xticks(range(df['domain_copy'].nunique()))
ax.set_xticklabels(df['domain_copy'].astype('category').cat.categories, rotation=0)
ax.set_yticks(range(df['kr_year_5'].nunique()))
ax.set_yticklabels(df['kr_year_5'].astype('category').cat.categories)
# ax.set_xlabel('기술 카테고리')
# ax.set_ylabel('갯수')
# ax.set_zlabel('기술 간극')

ax.bar3d(xpos, ypos, zpos, dx, dy, dz, alpha = 0.7, color=colors, shade=True)
ax.view_init(elev=45, azim=35)

ax.set_box_aspect((1, 1, 0.6))
ax.zaxis._axinfo["grid"]['linewidth'] = 0.5

legend_elements = [
    Patch(facecolor=domain_colors[d], label=d)
    for d in domain_cats
]
ax.legend(handles=legend_elements,
    title='기술 분야',
    loc='upper left',
    bbox_to_anchor=(1.02, 1))
fig.subplots_adjust(right=0.75)

# plt.tight_layout()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_subplot(projection='3d')

# 1. 고정된 순서 정의 (막대 위치와 색상을 일치시키기 위함)
target_domains = ['해상/해군', '항공/공군', '통신/전자', '육상/지상', '우주', '신기술', '미사일/방공']
year_bins = sorted(df['kr_year_5'].unique())

# 2. 모든 조합에 대해 미리 0으로 채워진 데이터 구조 생성 (데이터 누락 방지)
xpos, ypos, dz, colors = [], [], [], []

for ix, domain in enumerate(target_domains):
    for iy, year in enumerate(year_bins):
        # 해당 도메인과 연도에 맞는 데이터 개수 계산
        count = len(df[(df['domain'] == domain) & (df['kr_year_5'] == year)])
        
        xpos.append(ix)
        ypos.append(iy)
        dz.append(count)
        colors.append(domain_colors[domain]) # 딕셔너리에서 직접 색상 추출

xpos = np.array(xpos)
ypos = np.array(ypos)
zpos = np.zeros_like(xpos)
dx = dy = 0.7  # 막대 굵기 조절

# 3. 그래프 그리기
ax.bar3d(xpos, ypos, zpos, dx, dy, dz, color=colors, alpha=0.7, shade=True)

# 4. 축 라벨 설정 (순서 고정)
ax.set_xticks(range(len(target_domains)))
ax.set_xticklabels(target_domains, rotation=0) # 글자 겹치면 회전
ax.set_yticks(range(len(year_bins)))
ax.set_yticklabels(year_bins)

legend_elements = [
    Patch(facecolor=domain_colors[d], label=d)
    for d in domain_cats
]
ax.legend(handles=legend_elements,
    title='기술 분야',
    loc='upper right',
    bbox_to_anchor=(1.02, 1))
fig.subplots_adjust(right=0.75)

ax.view_init(elev=45, azim=35)
plt.show()
```

2. gap_years 비교를 통해 우리나라가 밀고있는 산업군 top3

```python
lineup = df['gap_years'].sort_values(ascending = False)
lineup2 = df['gap_years'].sort_values(ascending = True)
long10 = lineup.head(10)
short10 = lineup2.head(10)
# print(df[df['gap_years'] == 52.5])
# print(df[df['gap_years'] == 49.5])
# print(df[df['gap_years'] == 47.5])

lgaps = df[df['gap_years'].isin(long10)]
sgaps = df[df['gap_years'].isin(short10)]

l = lgaps['domain'].value_counts()
s = sgaps['domain'].value_counts()

plt.figure(figsize = (10, 5))

plt.subplot(1, 2, 1)
x = np.arange(len(l))
plt.bar(x, l.values, color = ['#ff7f0e', '#2ca02c', '#e377c2', '#1f77b4'])
# plt.text(1.0, 0.5, l)
plt.xticks(x, l.index, rotation = 45)
plt.title('기술 간극 큰 연구 목록')

plt.subplot(1, 2, 2)
x = np.arange(len(s))
plt.bar(x, s.values, color = ['#ff7f0e', '#2ca02c', '#9467bd', '#1f77b4'])
# plt.text(1.0, 0.7, s)
plt.xticks(x, s.index, rotation = 45)
plt.title('기술 간극 작은 연구 목록')

plt.tight_layout()
plt.show()
```
