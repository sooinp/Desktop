```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import re
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
df = pd.read_csv('course_completion_raw_1000.csv')
df
```

```python
print('user_id |', type(df['user_id'][0]))
print('watch_ratio |', type(df['watch_ratio'][0]))
print('weekly_minutes |', type(df['weekly_minutes'][0]))
print('login_days_7d |', type(df['login_days_7d'][0]))
print('sessions_7d |', type(df['sessions_7d'][0]))
print('assignment_submit_rate |', type(df['assignment_submit_rate'][0]))
print('quiz_avg |', type(df['quiz_avg'][0]))
print('forum_posts_30d |', type(df['forum_posts_30d'][0]))
print('device_main |', type(df['device_main'][0]))
print('signup_channel |', type(df['signup_channel'][0]))
print('paid |', type(df['paid'][0]))
print('signup_date |', type(df['signup_date'][0]))
print('last_active_date |', type(df['last_active_date'][0]))
print('completed |', type(df['completed'][0]))
```

```python
df['watch_ratio'] = df['watch_ratio'].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df['watch_ratio'] = df['watch_ratio'].astype(float)
df.loc[df['watch_ratio'] > 1, 'watch_ratio'] = df['watch_ratio'] / 100
df['watch_ratio'] = df['watch_ratio'].round(2)
# df['watch_ratio']
```

```python
df['weekly_minutes'] = df['weekly_minutes'].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df['weekly_minutes'] = df['weekly_minutes'].astype(float)
df['weekly_minutes'] = df['weekly_minutes'].round(0).astype(int) #정수 반올림(분)
# df['weekly_minutes']
```

```python
a = df['login_days_7d'].unique()
print(sorted(a))
b = df['sessions_7d'].unique()
print(sorted(b))
c = df['forum_posts_30d'].unique()
print(sorted(c))
```

```python
df['assignment_submit_rate'] = df['assignment_submit_rate'].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df['assignment_submit_rate'] = df['assignment_submit_rate'].astype(float)
df.loc[df['assignment_submit_rate'] > 1, 'assignment_submit_rate'] = df['assignment_submit_rate'] / 100
df['assignment_submit_rate'] = df['assignment_submit_rate'].round(2)
# df['assignment_submit_rate']
```

```python
df['quiz_avg'] = df['quiz_avg'].astype(str).str.replace(r'[^0-9.]', '', regex=True).str.strip()
df['quiz_avg'] = pd.to_numeric(df['quiz_avg'], errors='coerce')
df['quiz_avg'] = df['quiz_avg'].round(0)
df['quiz_avg']
```

```python
d = df['device_main'].unique()
print(sorted(d))

d_dic = {
    ' MOBILE': 'mobile',
    ' mobile': 'mobile',
    'MOBILE': 'mobile',
    'mobile': 'mobile',
    ' PC': 'pc',
    ' pc': 'pc',
    'PC': 'pc',
    'pc': 'pc',
    ' tablet': 'tablet',
    'TABLET': 'tablet',
    'tablet': 'tablet',
}
df['device_main'] = df['device_main'].replace(d_dic)
d = df['device_main'].unique()
print(sorted(d))
```

```python
df['signup_channel'].unique()
```

```python
d_dic2 = { # 0 = True, 1 = False임
    '0': 0,
    'True': 0,
    'Y': 0,
    '1': 1,
    'False': 1,
    'N': 1,
}
df['paid'] = df['paid'].replace(d_dic2)

print(df['paid'].unique())
print()
print(df[df['paid'].isna()])
print("\n결측치 개수:", df['paid'].isna().sum())
df = df.dropna(subset=['paid']) #결측치 아예 제거함
```

```python
df['signup_date'].unique()
```

```python
df['last_active_date'].unique()
```

```python
d_dic3 = {
    '중도 포기': 0,
    '완강': 1,
}
df['completed'] = df['completed'].replace(d_dic3)
print(df['completed'].unique())
```

```python
date_cols = ['signup_date', 'last_active_date']

for col in date_cols:
    # errors='coerce'를 설정하면 잘못된 날짜 형식은 NaT(결측치)로 처리되어 안전
    df[col] = pd.to_datetime(df[col], errors='coerce')

print(df[date_cols].info())
print(df[date_cols].head())
```

```python
df
```

```python
drop_cols = ['user_id', 'device_main', 'signup_channel', 'signup_date', 'last_active_date']
df_ml = df.drop(columns=drop_cols)

X = df_ml.drop(columns=['completed']).astype(float) # 타겟 제외 모두 X에 집어넣음
n_features = X.shape[1]
y = df_ml['completed']

model = Sequential()
model.add(Dense(12, input_dim = n_features, activation = 'relu', name = 'Dense_1'))
model.add(Dense(8, activation = 'relu', name = 'Dense_2'))
model.add(Dense(1, activation = 'sigmoid', name = 'Dense_3'))
model.summary()

model.compile(loss = 'binary_crossentropy', optimizer = 'adam', metrics = ['accuracy'])
history = model.fit(X, y, epochs = 100, batch_size = 5)
```
