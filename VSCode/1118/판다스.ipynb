1. 시리즈Series

- 1차원 배열의 값에 각 값에 대응되는 인덱스를 부여할 수 있는 구조
- 시리즈 = 값 + 인덱스

- 인덱스 길이 == 데이터 길이
- 인덱스 값은 문자열, 시간, 날짜, 정수 등도 모두 가능

- 리스트 = 파이썬 기본 자료구조  VS. 시리즈 = 판다스의 1차원 자료구조

### 연습 문제 ###

```python
#1.
#(1) 임의로 두 개의 시리즈 객체를 만듦. 모두 문자열 인덱스를 가져야 하며, 두 시리즈에 공통적으로 포함되지 않는 라벨 있어야 함
#(2) 위에서 만든 두 시리즈 객체 이용해 사칙 연산

k1 = pd.Series({"백": 100, "이백": 200, "삼백": 300, "사백": 400}, index=["백", "이백", "삼백", "사백"])
k2 = pd.Series({"오백": 500, "육백": 600, "칠백": 700, "팔백": 800}, index=["오백", "육백", "칠백", "팔백"])

ressum = k1["백"] + k2["오백"]
resmin = k1["이백"] - k2["육백"]
resmul = k1["삼백"] * k2["칠백"]
resdiv = k1["사백"] / k2["팔백"]

print(ressum, resmin, resmul, resdiv)
```

```python
cost = pd.Series([1500, 5000, 3500, 3000, 1000], index=["칫솔", "고무장갑", "포카칩", "베이킹소다", "스펀지"])
r = pd.Series([5, 1, 3, 2, 5], index=["칫솔", "고무장갑", "포카칩", "베이킹소다", "뒤집개"])

res = cost * r
res
```

3-2) 열 데이터의 갱신, 추가, 삭제

- 데이터프레임은 열 시리즈의 딕셔너리로 볼 수 있으므로 열 단위로 데이터를 갱신, 추가, 삭제 가능

```python
#"2010-2015 증가율" 이라는 이름의 열 추가

df2["2010~2015 증가율"] = df2["2010-2015 증가율"] * 100
df2
```

```python
# "2005-2010 증가율"이라는 이름의 열 추가
df2["2005-2010 증가율"] = ((df2["2010"] - df2["2005"]) / df2["2005"] * 100).round(3)
df2
```

```python
#"2010~2015 증가율"이라는 이름의 열 삭제
del df2["2010~2015 증가율"]
df2
```

```python
#하나의 열만 인덱싱하면 시리즈 반환
df2["지역"]
```
